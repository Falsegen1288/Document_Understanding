# Visual Table Extraction Comparison
This notebook compares the two table structure recognition models:
1. **Docling + TableFormer** (Deep structure & cell span restorer)
2. **Table Transformer (TATR)** (Object detection cell/grid classifier)

The comparison shows drawn cell grid overlays and side-by-side parsed DataFrames.

```python
import os
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt

pdf_path = 'data/scientific/Scientific_001.pdf'
img_path = 'data/scientific/Scientific_001.png'

print("PDF status:", os.path.exists(pdf_path))
print("Page image status:", os.path.exists(img_path))
```

```python
# 1. Run Docling table extraction
from algorithms.table_extraction.docling_tableformer.extractor import extract_tables as docling_extract

if os.path.exists(pdf_path):
    print("Running Docling table extraction...")
    docling_tables = docling_extract(pdf_path, pages=[1])
    print(f"Docling found {len(docling_tables)} tables.")
    if docling_tables:
        df = pd.DataFrame(docling_tables[0].get('data', []))
        print(df.head())
else:
    print("Docling cannot run without pdf_path.")
```

```python
# 2. Run TATR table extraction
from algorithms.table_extraction.tatr.extractor import extract_tables as tatr_extract

if os.path.exists(img_path):
    img = Image.open(img_path)
    # We crop a table region (e.g. coordinates [100, 600, 1100, 1200])
    t_bbox = [100, 600, 1100, 1200]
    cropped_table = img.crop(t_bbox)
    
    print("Running TATR on crop...")
    tatr_tables = tatr_extract(cropped_table)
    print(f"TATR found {len(tatr_tables)} tables.")
    if tatr_tables:
        df = pd.DataFrame(tatr_tables[0].get('data', []))
        print(df.head())
        
    plt.imshow(cropped_table)
    plt.axis('off')
    plt.show()
else:
    print("TATR cannot run without page image crop.")
```
