# Visual Figure Captioning & VLM Comparison
This notebook compares output diagrams captions generated by different VLMs:
1. **Groq Cloud API**: Llama 4 Scout 17B instruction-tuned VLM and Qwen 3.6 27B
2. **Local Ollama VLM**: Qwen2.5-VL 3B and Moondream

Medical diagram crops are shown with captions rendered side-by-side.

```python
import os
from PIL import Image
import matplotlib.pyplot as plt

img_path = 'data/scientific/Scientific_001.png'
if not os.path.exists(img_path):
    print("Place Scientific_001.png in data/scientific/")
else:
    img = Image.open(img_path)
    # Crop a diagram element from coordinates [150, 870, 1120, 1545]
    crop_box = [150, 870, 1120, 1545]
    cropped_fig = img.crop(crop_box)
    plt.imshow(cropped_fig)
    plt.axis('off')
    plt.show()
```

```python
from algorithms.image_extraction.groq.extractor import describe_figure as groq_desc
from algorithms.image_extraction.local.extractor import describe_figure as local_desc

prompt = "Describe this product image in one or two sentences, focusing on what the item is, its material/appearance, and any visible identifying features."

print("1. Requesting Groq Llama 4 Scout...")
llama_res = groq_desc(cropped_fig, prompt=prompt, model="meta-llama/llama-4-scout-17b-16e-instruct")
print("Groq Llama 4:", llama_res.get('description', ''))

print("\n2. Requesting Groq Qwen 3.6...")
qwen_g_res = groq_desc(cropped_fig, prompt=prompt, model="qwen/qwen3.6-27b")
print("Groq Qwen:", qwen_g_res.get('description', ''))

print("\n3. Requesting Local Ollama Qwen2.5-VL...")
qwen_l_res = local_desc(cropped_fig, prompt=prompt, model="qwen2.5vl:3b")
print("Local Qwen2.5-VL:", qwen_l_res.get('description', ''))

print("\n4. Requesting Local Ollama Moondream...")
moon_res = local_desc(cropped_fig, prompt=prompt, model="moondream:latest")
print("Local Moondream:", moon_res.get('description', ''))
```
